# Vascular Function Curve and Cardiac Function Curve





## Background

Cardiac output and venous return are fundamental variables in cardiovascular physiology. Cardiac output is the volume of blood pumped by the heart per minute, while venous return is the volume of blood returning to the right atrium per minute. In a steady state, these two flows must be equal:

$$
CO = VR
$$

Cardiac output can also be expressed as the product of heart rate and stroke volume:

$$
CO = HR \cdot SV
$$

The cardiac function curve, also known as the Frank–Starling curve, describes the relationship between cardiac filling and cardiac output. Right atrial pressure, $P_{RA}$, is commonly used as an approximation of cardiac preload. Preload is the degree to which the ventricular muscle fibres are stretched at the end of diastole, immediately before contraction.

According to the Frank–Starling mechanism, an increase in venous return increases ventricular filling and stretches the myocardial fibres. Within the physiological range, this improves the force of contraction and increases stroke volume. Consequently, cardiac output increases. This intrinsic mechanism allows the heart to adjust its output to changes in venous return and helps the right and left ventricles maintain equal outputs.

The cardiac function curve is plotted with right atrial pressure on the horizontal axis and cardiac output on the vertical axis. At low $P_{RA}$, the heart is underfilled and cardiac output is low. As $P_{RA}$ increases, cardiac output rises because ventricular preload increases. The curve is usually steep over the physiological range, indicating that relatively small changes in preload can produce substantial changes in cardiac output.

At very high values of $P_{RA}$, the cardiac function curve approaches a plateau. The myocardial fibres have reached, or exceeded, their optimal length, so further increases in preload produce little additional force or cardiac output. Excessively high filling pressures may contribute to venous and pulmonary congestion, particularly when cardiac function is impaired.

A simplified mathematical representation of the cardiac function curve is:

$$
CO = CO_{\max}
\left(
1-\exp\left[-\frac{P_{RA}-P_0}{k}\right]
\right)
$$

Here, $CO_{\max}$ is the maximum cardiac output predicted by the model, $P_0$ is a reference pressure, and $k$ determines the steepness of the curve. A smaller value of $k$ produces a steeper curve, meaning that cardiac output responds more strongly to changes in preload. During exercise, sympathetic stimulation increases myocardial contractility, which generally increases $CO_{\max}$ and changes the shape of the cardiac function curve.

The vascular function curve, also called the venous return curve, describes the relationship between right atrial pressure and venous return. It is plotted with $P_{RA}$ on the horizontal axis and venous return on the vertical axis. Unlike the cardiac function curve, the vascular function curve normally slopes downward.

The main driving pressure for venous return is the pressure difference between mean systemic filling pressure, $P_{MS}$, and right atrial pressure:

$$
VR = \frac{P_{MS}-P_{RA}}{R_S}
$$

In this equation, $R_S$ is the systemic resistance to venous return. Mean systemic filling pressure is the pressure that would exist throughout the systemic circulation if the heart stopped and blood were allowed to redistribute. It depends mainly on blood volume, venous tone, and vascular compliance.

Increasing $P_{RA}$ reduces the pressure gradient driving blood from the systemic circulation into the right atrium. Venous return therefore decreases. When $P_{RA}=P_{MS}$, the pressure gradient is zero and venous return becomes zero. At very low or negative values of $P_{RA}$, the large veins may collapse, limiting further increases in venous return and producing a plateau in the vascular function curve.

When arterial and venous compliance are included in the model, venous return may be represented as:

$$
VR =
\frac{1+C_v/C_a}{R_S}
\left(P_{MS}-P_{RA}\right)
$$

Here, $C_a$ is arterial compliance and $C_v$ is venous compliance. The expression determines the slope of the vascular function curve. With $C_a$ and $C_v$ in L/mmHg and $R_S$ in mmHg·min/L, as in this notebook, the flow is obtained directly in $\mathrm{L/min}$.

Blood volume and venous tone primarily affect the horizontal position of the vascular function curve by changing $P_{MS}$. Blood loss reduces the stressed vascular volume and lowers $P_{MS}$. The vascular function curve therefore shifts to the left, resulting in a lower venous return at a given $P_{RA}$. Venoconstriction increases venous tone and transfers blood from the unstressed to the stressed vascular volume. This raises $P_{MS}$ and shifts the curve to the right. Venodilation has the opposite effect.

Systemic resistance affects the slope of the vascular function curve. A lower resistance allows blood to return to the heart more easily and makes the curve steeper. Vascular compliance also influences the curve because changes in compliance alter the distribution of blood between the stressed and unstressed vascular compartments.

When the cardiac and vascular function curves are plotted together, their intersection represents the operating point of the cardiovascular system. At this point, cardiac output equals venous return, and both the steady-state flow and right atrial pressure are determined. Changes in myocardial contractility shift the cardiac function curve, whereas changes in blood volume, venous tone, vascular resistance, or vascular compliance alter the vascular function curve.

These curves are used to analyse the interaction between the heart and the systemic circulation. Cardiac function curves are useful for examining preload, contractility, exercise, and heart failure. Vascular function curves are useful for examining blood loss, venoconstriction, venodilation, systemic resistance, and vascular compliance. Together, they provide a graphical framework for explaining how physiological changes affect cardiac output, venous return, and right atrial pressure.

```python
# Cardiac Function Curve
import numpy as np

# Frank–Starling (exponential saturation)
def CO_starling(P: np.ndarray, CO_max=7.0, P0=-2.0, k=3.0) -> np.ndarray:
    CO = CO_max * (1 - np.exp(-(P - P0)/k))
    return np.maximum(CO, 0.0)
```

```python

import numpy as np
import matplotlib.pyplot as plt

rest_co = dict(CO_max=8.0, P0=-3, k=2.5)   # contractility/Starling shape
ex_co = dict(CO_max=15.0, P0=-3, k=1.2)

P_RA = np.linspace(-3, 10, 500)

CO_rest = CO_starling(P_RA, rest_co["CO_max"], rest_co["P0"], rest_co["k"])
CO_ex = CO_starling(P_RA, ex_co["CO_max"], ex_co["P0"], ex_co["k"])


plt.plot(P_RA, CO_rest,label="rest")
plt.plot(P_RA, CO_ex,label="exercise")
plt.xlabel("Right atrial pressure (mmHg)")
plt.ylabel("Cardiac Output (L/min)")
plt.title("Cardiac Function Curve")
plt.legend()
plt.show()
```


## Vascular Function Curve

The **vascular function curve** is a graphical representation that shows how **venous return** changes as a function of **right atrial pressure ($P_{RA}$)**.
It's a key concept in cardiovascular physiology because it illustrates the relationship between **systemic filling pressure**, **venous resistance**, and **cardiac preload**.

*   **X-axis:** Right atrial pressure ($P_{RA}$), which acts as the back-pressure against venous return.
*   **Y-axis:** Venous return (or cardiac output in a closed system).
*   The curve typically slopes downward

# Steady State Operating Point between Cardiac and Vascular Function Curves

## Background

The cardiovascular system operates as an integrated pump-and-circuit system. The heart generates cardiac output, while the systemic circulation determines how much blood can return to the heart. Cardiac output and venous return are therefore not independent variables. In a steady state, the volume pumped by the heart must equal the volume returning to the right atrium:

$$
CO = VR
$$

The **cardiac function curve (CFC)** describes the relationship between right atrial pressure, $P_{RA}$, and cardiac output, $CO$. Right atrial pressure is used as an estimate of cardiac preload, which is the degree of ventricular filling and myocardial stretch at the end of diastole. Through the Frank–Starling mechanism, an increase in preload initially increases the force of ventricular contraction, stroke volume, and cardiac output.

A simplified cardiac function curve can be written as:

$$
CO(p_{RA}) =
CO_{max}
\left[
1-\exp\left(
-\frac{P_{RA}-P_0}{k}
\right)
\right]
$$

Here, $CO_{\max}$ is the maximum cardiac output predicted by the model, $P_0$ is a reference pressure, and $k$ determines how rapidly the curve approaches its maximum. The curve is increasing over the physiological range and eventually approaches a plateau. This plateau reflects the fact that the myocardium cannot increase its force of contraction indefinitely as preload rises.

Cardiac contractility changes the position and shape of the cardiac function curve. Increased sympathetic stimulation, such as during exercise, increases contractility and generally shifts the curve upward. The heart can then generate a greater cardiac output at the same $P_{RA}$. Reduced contractility, as observed in systolic heart failure, shifts the curve downward and reduces the maximum cardiac output.

The **vascular function curve (VFC)** describes the relationship between right atrial pressure and venous return. It normally slopes downward because $P_{RA}$ acts as back-pressure opposing the flow of blood from the systemic circulation into the right atrium. The pressure driving venous return is the difference between mean systemic filling pressure, $P_{MS}$, and right atrial pressure:

$$
VR =
\frac{P_{MS}-P_{RA}}{R_S}
$$

In this relationship, $R_S$ represents the effective resistance to venous return. Mean systemic filling pressure is the pressure that would be present throughout the systemic circulation if the heart stopped and blood flow were allowed to redistribute. It depends mainly on blood volume, venous tone, and vascular compliance.

When $P_{RA}$ increases, the pressure gradient for venous return decreases, and venous return falls. When $P_{RA}=P_{MS}$, the pressure gradient is zero and the predicted venous return is zero. At very low right atrial pressures, the large veins may collapse, which limits further increases in venous return and produces a plateau in the vascular function curve.

In the model used in this exercise, arterial and venous compliance contribute to the effective slope of the vascular function curve:

$$
VR =
\frac{1+C_v/C_a}{R_S}
\left(P_{MS}-P_{RA}\right)
$$

Here, $C_a$ is arterial compliance and $C_v$ is venous compliance. Depending on the units of the parameters, the calculated flow may be expressed in $\mathrm{mL/min}$ and must be divided by $1000$ to obtain $\mathrm{L/min}$.

The position of the vascular function curve is strongly influenced by $P_{MS}$. An increase in blood volume or venous constriction increases the stressed vascular volume and raises $P_{MS}$. This shifts the vascular function curve to the right, increasing venous return at a given $P_{RA}$. Conversely, haemorrhage or venodilation reduces stressed vascular volume, lowers $P_{MS}$, and shifts the curve to the left.

The slope of the vascular function curve is influenced by systemic resistance and vascular compliance. A lower effective resistance increases the amount of blood that can return for a given pressure gradient, producing a steeper curve. A higher resistance reduces venous return and produces a flatter curve. Changes in compliance alter how blood is distributed between the stressed and unstressed vascular compartments and can therefore affect both the slope and position of the curve.

The **steady-state operating point** is the intersection of the cardiac and vascular function curves. At this point, the cardiac output generated by the heart is exactly equal to the venous return provided by the systemic circulation:

$$
CO(P_{RA}) = VR(P_{RA})
$$

The corresponding values of $P_{RA}$ and flow represent the equilibrium state of the cardiovascular system. The operating point therefore provides an estimate of the person’s steady-state cardiac output, venous return, and right atrial pressure.

The operating point can be determined graphically by plotting both curves on the same axes and identifying their intersection. It can also be calculated numerically by finding the value of $P_{RA}$ for which the difference between the two curves is zero:

$$
CO(P_{RA})-VR(P_{RA})=0
$$

Changes affecting the heart or the systemic circulation move the operating point. Increased contractility shifts the cardiac function curve upward and usually increases cardiac output while reducing the equilibrium value of $P_{RA}$. Increased blood volume or venoconstriction shifts the vascular function curve to the right and generally increases both cardiac output and right atrial pressure. Haemorrhage shifts the vascular curve to the left, causing cardiac output and right atrial pressure to decrease.

During exercise, increased sympathetic activity raises cardiac contractility and usually increases heart rate. The cardiac function curve therefore shifts upward, allowing a higher cardiac output. During transfusion, the increased blood volume raises $P_{MS}$ and shifts the vascular function curve to the right. During vasoconstriction, venous tone increases and systemic resistance may also change, affecting both the position and slope of the vascular function curve.

Studying the two curves together is useful because it shows that cardiac output is determined by the interaction between cardiac pumping ability and venous return. The cardiac function curve describes what the heart can pump at different filling pressures, whereas the vascular function curve describes how much blood the circulation can return at those pressures. Their intersection summarizes the combined cardiovascular response to exercise, transfusion, haemorrhage, vasoconstriction, vasodilation, changes in resistance, and changes in compliance.

````python
import numpy as np
from scipy.optimize import brentq

# Frank–Starling (exponential saturation)
def CO_starling(P: np.ndarray|float,*, CO_max=8.0, P0=-3.0, k=3.0) -> np.ndarray:
    CO = CO_max * (1 - np.exp(-(P - P0)/k))
    return CO

# Vascular Function Curve
def vascular_function_curve(P_RA: np.ndarray|float,*, P_MS: float, R_S: float, C_a:float, C_v:float) -> np.ndarray|float:
    """Calculate venous return based on right atrial pressure, mean systemic pressure, and systemic resistance."""
    factor = (1+C_v / C_a) / R_S
    CO = -factor * (P_RA - P_MS)
    return CO


def operating_point(P_RA: np.ndarray, co_params: dict, vascular_params: dict):
    def CO_VR_difference(p):
        CO = CO_starling(p, **co_params)
        VR = vascular_function_curve(p, **vascular_params)
        return CO - VR
    P_RA_op = brentq(CO_VR_difference, P_RA[0], P_RA[-1])
    assert isinstance(P_RA_op, float), "Operating point should be a float" # Jeg added this assertion to ensure that P_RA_op is a float
    CO_op = CO_starling(P_RA_op, **co_params)
    return P_RA_op, CO_op
````
